```python

import pandas as pd

data = {
    "Product": ["Laptop", "Mobile", "Tablet", "Laptop", "Mobile",
                "Tablet", "Laptop", "Mobile", "Tablet", "Laptop"],
    "Category": ["Electronics"] * 10,
    "Sales": [50000, 20000, 15000, 55000, 22000,
              18000, 48000, 25000, 16000, 60000],
    "Quantity": [2, 5, 3, 2, 4, 3, 1, 6, 4, 2],
    "City": ["Chennai", "Coimbatore", "Madurai", "Chennai",
             "Salem", "Madurai", "Coimbatore", "Chennai",
             "Salem", "Chennai"]
}

df = pd.DataFrame(data)

print("Dataset created successfully!")
print(df)
```

```python

# 1. Dataset-la ethana rows and columns irukku?
print("Dataset Shape:", df.shape)

# 2. Column names check pannuvom
print("\nColumn Names:")
print(df.columns)

# 3. Missing values check pannuvom
print("\nMissing Values:")
print(df.isnull().sum())

# 4. Duplicate rows check pannuvom
print("\nDuplicate Rows:", df.duplicated().sum())

# 5. Sales details
print("\nSales Statistics:")
print(df["Sales"].describe())
```

```python

# Calculate total sales for each product
product_sales = df.groupby("Product")["Sales"].sum()

print("Total Sales by Product:")
print(product_sales)

# Find the product with highest sales
print("\nHighest Sales Product:")
print(product_sales.idxmax())

# Find the product with lowest sales
print("\nLowest Sales Product:")
print(product_sales.idxmin())
```

```python

import matplotlib.pyplot as plt

product_sales = df.groupby("Product")["Sales"].sum()

product_sales.plot(kind="bar")

plt.title("Product-wise Total Sales")
plt.xlabel("Product")
plt.ylabel("Total Sales (₹)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

city_sales = df.groupby("City")["Sales"].sum().sort_values(ascending=False)

city_sales.plot(kind="bar")

plt.title("City-wise Total Sales")
plt.xlabel("City")
plt.ylabel("Total Sales (₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

print("EDA FINDINGS")
print("-" * 30)

print("Total Sales:", df["Sales"].sum())
print("Average Sales:", df["Sales"].mean())
print("Highest Sales:", df["Sales"].max())
print("Lowest Sales:", df["Sales"].min())
print("Total Quantity Sold:", df["Quantity"].sum())

print("\nCity-wise Sales:")
print(df.groupby("City")["Sales"].sum().sort_values(ascending=False))

print("\nProduct-wise Sales:")
print(df.groupby("Product")["Sales"].sum().sort_values(ascending=False))
```

```python

product_sales = df.groupby("Product")["Sales"].sum()

plt.figure(figsize=(7, 7))
plt.pie(
    product_sales,
    labels=product_sales.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Product-wise Sales Distribution")
plt.show()
```

```python

quantity_sales = df.groupby("Product")["Quantity"].sum()

quantity_sales.plot(kind="bar")

plt.title("Quantity Sold by Product")
plt.xlabel("Product")
plt.ylabel("Total Quantity Sold")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

city_sales = df.groupby("City")["Sales"].sum()

plt.figure(figsize=(7, 7))
plt.pie(
    city_sales,
    labels=city_sales.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("City-wise Sales Distribution")
plt.show()
```

```python

from bs4 import BeautifulSoup
import requests
import pandas as pd

url = "https://books.toscrape.com/"
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

books = soup.find_all("article", class_="product_pod")

book_data = []

for book in books:
    title = book.h3.a["title"]
    price = book.find("p", class_="price_color").text
    availability = book.find(
        "p", class_="instock availability"
    ).text.strip()

    book_data.append({
        "Book Title": title,
        "Price": price,
        "Availability": availability
    })

df_books = pd.DataFrame(book_data)

print("Books extracted:", len(df_books))
print(df_books.head(10))
```

```python

# Save scraped book data as a CSV file
df_books.to_csv("CodeAlpha_WebScraping.csv", index=False)

print("CSV file saved successfully!")
print("Total books saved:", len(df_books))
```
